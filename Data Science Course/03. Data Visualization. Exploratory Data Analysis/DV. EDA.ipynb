```python
# Cleaner code
segments = {}
for segment, group_data in electric_vehicles.groupby("segment"):
    segments[segment] = group_data.efficiency_wh_per_km.values
```

```python
plt.boxplot(segments.values())
plt.xticks(rotation = 90)
plt.gca().set_xticklabels(segments.keys())
plt.show()
```

```python
drive_trains = {}
for drive_train, group_data in electric_vehicles.groupby("drivetrain"):
    drive_trains[drive_train] = group_data.efficiency_wh_per_km.values
```

```python
# plt.boxplot(drive_trains.values())
# plt.xticks(rotation = 90)
# plt.gca().set_xticklabels(drive_trains.keys())
# plt.show()
# Convert dictionary values to a list and keep track of keys for labels
values = list(drive_trains.values())
keys = list(drive_trains.keys())

# Create the boxplot with explicit positions
plt.boxplot(values)

# Set the x-tick positions to match the number of boxes
# and use the keys as labels
plt.xticks(range(1, len(keys) + 1), keys, rotation=90)
plt.ylabel("Efficiency, Wh/km")
# Bigger number -> bigger efficiency
# Possible question: how different are those medians and do a T-test
# Can be done with violinplot using cborn library

plt.show()
```
