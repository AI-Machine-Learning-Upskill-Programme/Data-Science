```python
electric_vehicles[electric_vehicles.cargo_volume_l.fillna("").str.contains("banana", case = False)]
```

```python
electric_vehicles["cargo_volume_l_corrected"] = electric_vehicles.cargo_volume_l.copy()
```

this will replace by reference; in order to avoid changes in the original column, I use copy()

```python
indices_to_correct = electric_vehicles[electric_vehicles.cargo_volume_l.fillna("").str.contains("banana", case = False)].index
```

```python
electric_vehicles.loc[indices_to_correct]
```

```python
electric_vehicles.loc[indices_to_correct, "cargo_volume_l_corrected"]
```

```python
electric_vehicles.loc[indices_to_correct, "cargo_volume_l_corrected"] = np.nan
```

```python
electric_vehicles.loc[indices_to_correct, "cargo_volume_l_corrected"]
```

```python
electric_vehicles.cargo_volume_l_corrected = electric_vehicles.cargo_volume_l_corrected.astype(float, errors = "ignore")
```

```python
plt.hist(electric_vehicles.cargo_volume_l_corrected, bins = 25)

plt.axvline(855.1688, c = "r") 

plt.xlabel("Volume [l]")
plt.ylabel("Count")
plt.show()
```

multimodal distribution

```python
base_value = 855.168 / 10 # 1 liter = ... source = ...
electric_vehicles.loc[indices_to_correct, "cargo_volume_l_corrected"] = [base_value * 10, base_value * 31, base_value * 13]
```

```python
electric_vehicles.loc[indices_to_correct, "cargo_volume_l_corrected"] = electric_vehicles.loc[indices_to_correct, "cargo_volume_l_corrected"].round()
```

```python
plt.hist(electric_vehicles.cargo_volume_l_corrected, bins = 25)

plt.axvline(855.1688, c = "r") 

plt.xlabel("Volume [l]")
plt.ylabel("Count")
plt.show()
```

the car at 2500 is an outlier
histograms help in searching for multimodality, often meaning system bias

```python
plt.hist(electric_vehicles.seats)
plt.show()
```

discrete distribution
Sometimes it is wrong to have discrete distribution

```python
electric_vehicles.seats.nunique()
```

```python
plt.hist(electric_vehicles.seats, bins = 20)
plt.show()
```

The pins on the numbers are at different places and that is an artefact because of the way matplotlib creates bins. The histograms matplotlib creates are for continuous variables. matplotlib threats the numbers at the bottom as labels such as the histogram about the brands (a categorical variable)

```python
electric_vehicles.seats.value_counts()
```

```python
plt.bar(electric_vehicles.seats.value_counts().index, electric_vehicles.seats.value_counts())

plt.title("The majority (~80%) of cars are 5-seaters")
plt.xlabel("Number of seats")
plt.ylabel("Count")
plt.show()
```
