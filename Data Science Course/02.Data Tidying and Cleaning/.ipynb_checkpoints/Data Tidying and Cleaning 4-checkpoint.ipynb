and does not work here instead of &
== has a bigger priority than the binary & - that is why everything between the conditions must be in () brackets

```python
billboard_data[
    (billboard_data.artist == "Creed") & 
    (billboard_data.track == "Higher") &
    (billboard_data.week >= 40)
]
```

```python
creed = billboard_data[
    (billboard_data.artist == "Creed") & 
    (billboard_data.track == "Higher") &
    (billboard_data.week >= 40)
]
```

```python
plt.plot(creed.week, creed.position)
```

```python
plt.show()
```

```python
plt.plot(creed.week, creed.position)
plt.ylim(1, 60)
plt.gca().invert_yaxis()
plt.show()
```

```python
creed = billboard_data[
    (billboard_data.artist == "Creed") & 
    (billboard_data.track == "Higher")
]
```

```python
plt.plot(creed.week, creed.position)
plt.ylim(1, 100)
plt.gca().invert_yaxis()
plt.show()
```
