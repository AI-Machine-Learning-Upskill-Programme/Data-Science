```python
# Define all replacement rules in one place
unit_map = {
    "msnm": "m",
    "m.s.n.m": "m",
    "masl": "m",
    "mts": "m",
    "mtrs": "m",
    "meter": "m",
    "meters": "m",
    "metros": "m",
    "公尺": "m",
    "pies": "ft",
    "pies.": "ft",
    "ft.": "ft",
    "feet": "ft",
    "ft": "ft",  # normalize even this
    "psn": "ft",
    "psnm": "ft",
    "p.s.n.m.": "ft",
    "p.s.n.m": "ft"
}

def normalize_units(value):
    for key, val in unit_map.items():
        value = value.replace(key, val)
    return value

def clean_altitude(value):
    if pd.isnull(value):
        return np.nan

    value = str(value).lower()

    # Normalize units
    value = normalize_units(value)

    # Clean approx and extra symbols
    value = re.sub(r"approx(?:\.|imately)?", "", value)
    value = re.sub(r"[~']", "", value)
    value = re.sub(r"[^0-9\.\-\sftm]", "", value)  # remove strange symbols

    # Extract numbers
    nums = re.findall(r"\d+(?:\.\d+)?", value)
    if not nums:
        return np.nan

    nums = [float(n) for n in nums]

    # Handle ranges (take average)
    avg = sum(nums) / len(nums)

    # Check if it's in feet
    if 'ft' in value:
        avg *= 0.3048  # Convert to meters

    return round(avg, 2)
```

```python
coffee_data.altitude = coffee_data.altitude.apply(clean_altitude)
```
