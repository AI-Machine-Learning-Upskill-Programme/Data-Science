```python
# Now I create a pivot table for each one of those countries to see what other measurement units they have
for value in filtered["country_of_origin"].unique():
    subset = coffee_data[coffee_data["country_of_origin"] == value][["country_of_origin","bag_weight"]]
    #print(subset)
    subset["bag_weight_num"] = subset["bag_weight"].str.extract(r"(\d+\.?\d*)").astype(float)
    
    pivot_count = subset.pivot_table(
        index="country_of_origin",
        columns = "bag_weight",
        values="bag_weight",
        aggfunc="count"
    )
    
    print(pivot_count)
    print("\n" + "-"*50 + "\n")
```

Unfortunately, a big mismatch has happened here - some values use kg as a measurement unit, some - lbs and others do not have a measurement unit at all.

Firstly, I will transform each and every value in lbs to kg because lbs is mostly used in the British Imperial and the USA.
    / source: https://en.wikipedia.org/wiki/Pound_(mass) /

```python
coffee_data["bag_weight_kg"] = coffee_data["bag_weight"]
lbs_to_kg = 0.45359237

# Create a mask for rows with "lbs" in bag_weight_kg
mask = coffee_data["bag_weight_kg"].str.contains("lbs", case=False, na=False) & \
       ~coffee_data["bag_weight_kg"].str.contains("kg", case=False, na=False)
numeric_values = (
    coffee_data.loc[mask, "bag_weight_kg"]
    .str.split()      # split by whitespace → ["60", "lbs"]
    .str[0]           # take first element ("60")
    .astype(float)    # convert to float
)
# coffee_data.loc[mask, "bag_weight"] = (numeric_values[0] * lbs_to_kg).round(2)
```

```python
numeric_values.value_counts(dropna = False)
```

```python
coffee_data.loc[mask, "bag_weight_kg"] = (numeric_values * lbs_to_kg).round()
coffee_data["bag_weight_kg"] = coffee_data["bag_weight_kg"].str.extract(r"(\d+)")
```

```python
# Convert to int so that only the whole part of the number can be used
```

```python
# Add kg to the numbers
coffee_data.loc[mask,"bag_weight"] = coffee_data.loc[mask, "bag_weight"].astype(str) + " kg"
```

```python
filtered = coffee_data.loc[
    ~coffee_data["bag_weight"].str.contains("kg", case=False, na=False),
    ["country_of_origin", "bag_weight"]
]
filtered["country_of_origin"].unique()
```

```python
# For each of these countries, a pivot table is created so that I can see the various values of bag weights 
for value in filtered["country_of_origin"].unique():
    subset = coffee_data[coffee_data["country_of_origin"] == value][["country_of_origin","bag_weight"]]
    #print(subset)
    subset["bag_weight_num"] = subset["bag_weight"].str.replace(" kg", "", regex=False).astype(float)
    
    pivot_count = subset.pivot_table(
        index="country_of_origin",
        columns = "bag_weight",
        values="bag_weight",
        aggfunc="count"
    )
    
    print(pivot_count)
    print("\n" + "-"*50 + "\n")

```

```python
country_weight_units_map = {
    "Guatemala": "kg",
    "United States (Hawaii)": "lbs",
    "Ethiopia": "kg",
    "Nicaragua": "kg",
    "Mexico": "kg"
}

# Create bag_weight column based on country_of_origin
filtered["bag_weight"] = (
    filtered["bag_weight"].astype(str) + " " +
    filtered["country_of_origin"].map(country_weight_units_map)
)

```
