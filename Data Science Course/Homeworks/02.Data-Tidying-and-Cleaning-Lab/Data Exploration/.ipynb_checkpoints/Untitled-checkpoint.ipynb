## Data Tidying and Cleaning

```python
# Reading the dataset... Naming it coffee_data
## The "../" helps referring to the parent folder of the current one.
coffee_data = pd.read_csv("../data/merged_data_cleaned.csv")
```

```python
coffee_data
```

```python
# Size of the dataset
coffee_data.shape
```

```python
# Columns (variables / features) in the dataset
coffee_data.columns
```

```python
# Types of variables
coffee_data.dtypes
```

```python
# Checking if the column "Unnamed: 0" is the same as the "index" column
coffee_data["Unnamed: 0"].nunique()
```

```python
# Remove the unnecessary column ["Unnamed: 0"] as it is the same as the index 
# (proven by the previous row - we have 1339 unique values which is exactly the role of the index)
coffee_data = coffee_data.drop(columns = "Unnamed: 0")
```

```python
coffee_data.columns
```

```python
# See if Species can be turned into a categorical variable instead of being an object var
coffee_data.Species
```

```python
coffee_data.Species.value_counts(dropna = False)
```

```python
# As there are only two types of coffee species, the feature "species" can be turned into a categorical variable
coffee_data.Species = coffee_data.Species.astype("category")
```

```python
coffee_data.info()
```

```python
# Make the columns in a more pythonic style
coffee_data.columns = coffee_data.columns.str.lower()
```

```python
coffee_data = coffee_data.rename(columns = lambda col: col.replace(".", "_"))
```

```python
coffee_data.species.info()
```

```python
coffee_data.columns
```

```python
coffee_data.owner
```

```python
# View the owners and how many times they appear in the dataset
coffee_data.owner.value_counts(dropna = False)
```

```python
# It is better to upper their names
coffee_data.owner = coffee_data.owner.str.upper()
```

```python
coffee_data.owner

coffee_data.owner.unique()
```

```python
# These are the records that do not have an owner but I spot something else here.
## A lot of records have a duplicate but with some little differences which come from the ico_number
coffee_data[coffee_data.owner.isna() == True]
```

```python
coffee_data[["owner", "farm_name"]].sample(20)
```

```python
# As I do not known the owner, I think that the best option is to replace the NaN values with "Unknown".
# According to my research, some coffee brands do not have an exact owner such as the coffee from Honduras - 
# actually there is no farm Ios Hicaques because that is a coffee-producing region in Honduras
coffee_data.owner = coffee_data.owner.fillna("Unknown")
```

```python
coffee_data[coffee_data.owner == "Unknown"]
```

```python
coffee_data["country_of_origin"].value_counts(dropna = False)
```

```python
coffee_data[coffee_data["country_of_origin"].isna()]
```

```python
# The country which is missing is Columbia according to my research
# source = "http://racafe.com.co/en/about-us/about-racafe/"
coffee_data["country_of_origin"] = coffee_data["country_of_origin"].fillna("Colombia")
```

```python
coffee_data.loc[1197]
```

```python
# I also spotted some errors in two other names - Côte d'Ivoire, and "Tanzania, United Republic of" is a little bit bland.

country_replacements = {
    "Tanzania, United Republic Of": "Tanzania",
    "Cote d?Ivoire": "Côte d'Ivoire"
}

coffee_data["country_of_origin"] = coffee_data['country_of_origin'].replace(country_replacements)
```

```python
coffee_data["country_of_origin"].value_counts(dropna = False)
```

```python
# Now I move on to the next column which is farm_name
coffee_data["farm_name"].info()
```

```python
# There are 359 NaN values in the "farm_name" column.
# That is why I will replace with "Unspecified"
1339 - 980
```

```python
coffee_data["farm_name"] = coffee_data["farm_name"].fillna("Unspecified")
```

```python
(coffee_data["farm_name"] == "Unspecified").value_counts()
```
