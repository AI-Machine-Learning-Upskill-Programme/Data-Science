### Problem 5. Dates (1 point)
This should remind you of problem 4 but it's slightly nastier. Fix the harvest years, document the process.

While you're here, fix the expiration dates, and grading dates. Unlike the other column, these should be dates (`pd.to_datetime()` is your friend).


```python
coffee_data["harvest_year"].value_counts(dropna = False)
```

What I will do here is to extract only the numerical part of the value. For example, for 4t/2011 I will leave only 2011.

```python
coffee_data["harvest_year"] = coffee_data["harvest_year"].str.extract(r"(19|20\d{2})")
coffee_data["harvest_year"].value_counts(dropna = False)
```

```python
coffee_data[coffee_data["harvest_year"].notna()][["harvest_year", "grading_date", "expiration"]]
```

```python
coffee_data[coffee_data["harvest_year"].isna()][["country_of_origin", "harvest_year", "grading_date", "expiration"]]
```

I can see that in most cases the harvest year is the yeaf before the year of the grading date. And the expiration is the year after the drading year.
However, for months such as January and December the harvest year is the same as the grading year.

Above I have found the countries, grading data and expiration corresponding to each missing harvest year.
Now for each of these countries, I will see if the grading year is later than the harvest year.

```python
coffee_data[coffee_data["country_of_origin"] == "Guatemala"][["harvest_year", "grading_date"]]
# Now I will compare the difference between each couple [grading year, harvest year]

# Convert grading_date to datetime
coffee_data["grading_date"] = pd.to_datetime(coffee_data["grading_date"], format = "mixed")
# Create a new feature called "grading_year"
coffee_data["grading_year"] = coffee_data["grading_date"].dt.year
```

```python
# Function to see the distribution of differences between grading year and harvest year in each country with missing harvest years.
def diff_distribution(country):
    # Take only the observations which have non-null harvest year values
    coffees_with_harvest_years = coffee_data[(coffee_data["harvest_year"].notna()) & (coffee_data["country_of_origin"] == country)][["country_of_origin", "harvest_year", "grading_year"]]
    differences = coffees_with_harvest_years["grading_year"].astype(int) - coffees_with_harvest_years["harvest_year"].astype(int)
    print(f"Country: {country}")
    print(differences.value_counts(normalize = True))
    
```

```python
for country in coffee_data[coffee_data["harvest_year"].isna()]["country_of_origin"].unique():
    diff_distribution(country)
    print("\n" + "-"*50 + "\n")
```

Since the difference of 0 prevails in most of the countries (56+ %), I will assign the grading year to the harvest years which are NaN.

```python
coffee_data["harvest_year"] = coffee_data["harvest_year"].fillna(coffee_data["grading_year"])
```

```python
# Last task from this problem: Convert expiration to datetime
coffee_data["expiration"] = pd.to_datetime(coffee_data.expiration, format = "mixed")
```

### Problem 9. Ratings (1 point)
The columns `Aroma`, `Flavor`, etc., up to `Moisture` represent subjective ratings. Explore them. Show the means and range; draw histograms and / or boxplots as needed. You can even try correlations if you want. What's up with all those ratings?

```python
coffee_data.info()
```

```python
def explore_ratings(feature):
    print(coffee_data[feature].describe())
```

```python
for col in coffee_data.columns[19:31]:
    #print(f"Feature name: {col}")
    print(explore_ratings(col))
    print("\n" + "-"*50 + "\n")
```

```python
cols = coffee_data.columns[19:31]

# Compute means and ranges
means = coffee_data[cols].mean()
ranges = coffee_data[cols].max() - coffee_data[cols].min()

# Plot means with ranges as error bars
plt.figure(figsize=(12,6))
plt.bar(cols, means, yerr=ranges/2, capsize=5, color='skyblue')
plt.xticks(rotation=45)
plt.ylabel('Value')
plt.title('Mean and Range of Selected Columns')
plt.show()

```

I can see that the means and ranges of all subjective rating except for 'total_cup_points' and 'moisture' are really close and they can actually be divided into two groups.
Aroma, Flavor, Aftertaste, Acidity, Body, Balance, Cupper points have nearly equal means and Uniformity, Clean cup, Sweatness form another group of almost equal means and ranges. Total cup points is a way higher than any other rating but this is because it shows total amount. It is interesting that moisture has quite low mean and range.
