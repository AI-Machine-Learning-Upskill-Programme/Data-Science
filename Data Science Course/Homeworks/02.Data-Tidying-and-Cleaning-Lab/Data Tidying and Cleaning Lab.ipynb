```python
import seaborn as sns
   
# Filter the companies with no missing measurement units in the countries which have missing m. units - 
# ['Guatemala', 'United States (Hawaii)', 'Ethiopia', 'Nicaragua','Mexico']
def create_pivot_and_bar_plot(country):
    
    # Filter companies for the specific country and keep needed columns
    companies = coffee_data[
        (coffee_data["country_of_origin"] == country) & 
        (coffee_data["measurement_unit"].notna())
    ][["country_of_origin", "owner", "measurement_unit"]]
    
    print(f"Country: {country}")
    
    # Create pivot table
    pivot_count = companies.pivot_table(
        index="owner",
        columns="measurement_unit",
        values="measurement_unit",
        aggfunc="count",
        observed=False
    )
    
    print(pivot_count)
    print()
    
    # Plot horizontal bar chart
    pivot_count.plot(kind='barh', stacked=True, figsize=(10, 8),
                     color={'kg':'skyblue', 'lbs':'lightcoral', 'kg lbs':'black'})
    
    plt.xlabel("Counts")
    plt.ylabel("Owner")
    plt.title(f"Measurement Unit by Owner in {country}")
    plt.legend(title='Unit', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tick_params(axis='y', labelsize = 7)
    plt.show()
    print("\n" + "-"*50 + "\n")
```

```python
# Function to see the distribution of differences between grading year and harvest year in each country with missing harvest years.
def diff_distribution(country):
    # Take only the observations which have non-null harvest year values
    coffees_with_harvest_years = coffee_data[(coffee_data["harvest_year"].notna()) & (coffee_data["country_of_origin"] == country)][["country_of_origin", "harvest_year", "grading_year"]]
    differences = coffees_with_harvest_years["grading_year"].astype(int) - coffees_with_harvest_years["harvest_year"].astype(int)
    print(f"Country: {country}")
    print(differences.value_counts(normalize = True))
    
```

```python
for country in coffee_data[coffee_data["harvest_year"].isna()]["country_of_origin"].unique():
    diff_distribution(country)
    print("\n" + "-"*50 + "\n")
```
