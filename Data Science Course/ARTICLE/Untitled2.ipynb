```python
from datasets import load_dataset
import os
from pathlib import Path

current_dir = Path(os.getcwd())
parent_dir = current_dir.parent
data_dir = parent_dir / "qasper-dataset"

dataset = load_dataset(
    "parquet",
    data_files={
        "train": os.path.join(data_dir, "train.parquet"),
        "validation": os.path.join(data_dir, "validation.parquet"),
        "test": os.path.join(data_dir, "test.parquet"),
    }
)

print(dataset)
```

```python
import subprocess
from pathlib import Path

current_dir = Path.cwd()
training_config = current_dir / "training_config" / "led_base_with_evidence_scaffold.jsonnet"
output_dir = current_dir / "output_dir"

try:
    result = subprocess.run(
        [
            "allennlp", "train", str(training_config),
            "-s", str(output_dir),
            "--include-package", "qasper_baselines"
        ],
        check=True,
        capture_output=True,  # capture stdout and stderr
        text=True             # return as strings
    )
except subprocess.CalledProcessError as e:
    print("Command failed!")
    print("Return code:", e.returncode)
    print("STDOUT:", e.stdout)
    print("STDERR:", e.stderr)
```
