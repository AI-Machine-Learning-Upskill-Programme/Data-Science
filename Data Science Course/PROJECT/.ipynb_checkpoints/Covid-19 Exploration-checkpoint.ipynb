##### Classification_final

Meaning: 
    covid test findings. Values 1-3 mean that the patient was diagnosed with covid in different
    degrees. 4 or higher means that the patient is not a carrier of covid or that the test is inconclusive.

```python
# Explore values
# According to the description of the feature in Kaggle the range of possible values is: [1; 3] --> covid  |   [4; +∞] --> not covid / unconvincing test
covid_data["classification_final"].unique()
```

Values 1 to 3: Indicate that the patient is diagnosed with COVID-19. The lower the value, the more severe or advanced the infection:
- 1 = Highest degree of infection (most severe case)
- 2 = Moderate degree of infection
- 3 = Mild degree of infection

Values 4 or higher: Indicate that the patient is either not infected with COVID-19 or that the test result is inconclusive.

For machine learning models, numerical values are preferred for categorical variables. However, for the purpose of this analysis, it is better to replace these numbers with descriptive words since the numeric values alone are not very meaningful.

```python
def replace_number(number):
    if number == 1:
        return "Severe COVID"
    elif number == 2:
        return "Moderate COVID"
    elif number == 3:
        return "Mild COVID"
    elif number >= 4:
        return "No COVID / Inconclusive"
    else:
        return "Invalid value"

covid_data["classification_final"] = [replace_number(num) for num in covid_data["classification_final"]]
```

```python
covid_data["classification_final"].value_counts(dropna = False, normalize = True)
```

Approx. 63% of the people are not ill from COVID or at least their COVID test is inconclusive.

```python
covid_data[covid_data["classification_final"] == "No COVID / Inconclusive"].sample(20)
```

And because I will need only those who are COVID cases for my analysis, I will drop all rows which have unclear diagnosis.

```python
covid_data = covid_data.drop(covid_data[covid_data["classification_final"] == "No COVID / Inconclusive"].index)
```

##### ICU (Intensive Care Unit)

Meaning: Indicates whether the patient had been admitted to an Intensive Care Unit.

```python
# Check values
covid_data.icu.value_counts(dropna = False)
```

```python
# In the description in Kaggle this is mentioned: In the Boolean features, 1 means "yes" and 2 means "no". values as 97 and 99 are missing data.
# Check the type of 'icu'
covid_data.icu.dtype
```

ICU status is not recorded as a boolean feature. Nevertheless, based on the description provided on Kaggle, icu can be treated as a boolean variable.

```python
covid_data.dtypes
```

There are no features recorded as boolean so this is a clue to search for the meaning of the variables rather than the type they belong to. If a variable can be treated as boolean, then I can convert it to boolean. Overall, this means that values 97 and 99 represent missing values for 'icu', 1 stands for "yes" (been in Intensive Unit Care) and 2 stands for "no" (Not been in Intensive Unit Care).

```python
# Convert 'icu' to type "category"
covid_data.icu = covid_data.icu.astype("category")
```

```python
# See which are the observations with missing values in 'icu' and what is the difference between missing value replaced with 97 and missing value replaced with 99
missing_values = covid_data[(covid_data.icu == 99)][["patient_type"]]
missing_values["patient_type"].value_counts(dropna = False)
```

```python
missing_values1 = covid_data[(covid_data.icu == 97)][["patient_type"]]
missing_values1["patient_type"].value_counts(dropna = False)
```

Now the difference is visible: 99 represents patients who have been hospitalized but it is uncertain if they were admited to an ICU, while 97 represents patients who were sent home but ICU status is uncertain.
Summary: 
- 97 = Sent home, ICU status uncertain
- 99 = Hospitalized, ICU status uncertain


```python
# I replace 97 and 99 with just one value for missing data, naming it "Unknown". The focus is only on the ICU status - yes, no or unknown;
# If information about the patient type is needed, the column "patient_type" can help so 97 and 99 are unnecessary for the analysis.

covid_data['icu'] = covid_data['icu'].cat.add_categories('Unknown')

covid_data['icu'] = covid_data['icu'].replace({
    1: 'Yes',
    2: 'No',
    97: 'Unknown',
    99: 'Unknown'
})
```

##### USMER

Meaning: USMER (Unidad de Salud Monitoras de Enfermedad Respiratoria / Viral Respiratory Disease Monitoring Health Unit)  
This indicates whether the patient treated medical units of the first, second, or third level. (Description on Kaggle, which is a little bit confusing) 

An explanation of the levels of healthcare in Mexico can be found here: https://www.internations.org/mexico-expats/guide/healthcare

- Free primary (e.g. General Practitioner or nurse) 
- Secondary (e.g. specialist doctor) healthcare. 
- Tertiary care (i.e. highly specialized treatments) is subject to fees.

Description of the feature after some research:
    Indicates whether the patient was treated in a sentinel surveillance health unit specialized in monitoring viral respiratory diseases (e.g., influenza, COVID-19). USMER units exist at all three levels of healthcare in Mexico:

- Value = 1 → Patient treated in a USMER unit (part of the national respiratory-disease surveillance network).
- Value = 2 → Patient treated in a non-USMER unit (outside the sentinel network).

```python
covid_data.usmer.value_counts(dropna = False)
```

```python
# The same procedures as the previous ones but this time there is no missing data.
covid_data.usmer = covid_data.usmer.astype("category")

covid_data.usmer = covid_data.usmer.replace({
        1: 'Yes',
        2: 'No',
})
```

## Questions which can lead to more hypothesis and discoveries

### Question 1: What is the most frequent pre-condition people who died from COVID had?

```python
# Select the columns with the pre-conditions
preconditions_columns = covid_data.columns[6:18]
covid_victims = covid_data[covid_data["date_died"].notna()][preconditions_columns]
covid_victims
```

```python
def count_preconditions():
    for col in covid_victims.columns:
        print(f"Pre-condition:")
        print(covid_victims[col].value_counts(dropna = False, normalize = True))
        print("\n" + "-"*50 + "\n")
count_preconditions()
```

From these statistics it can be seen that approx. 74% of those who died from COVID, had pneumonia.


### Question 2: What percentage of COVID-19 cases resulted in death?

```python
# Deaths count
deaths = covid_data["date_died"].notna().sum()

# Count of all infected people
total_cases = len(covid_data)

# Fatality (%)
fatality_rate = (deaths / total_cases) * 100

print(f"Overall fatality rate: {fatality_rate:.2f}%")
```

### Question 3: Which gender prevails in COVID cases? And in COVID deaths?

```python
sick_counts = covid_data.sex.value_counts(dropna = False)
sick_counts
```

```python
covid_victims["sex"] = covid_data[covid_data["date_died"].notna()].sex
death_counts = covid_victims.sex.value_counts(dropna = False)
death_counts
```

```python
df = pd.DataFrame({
    "Sick": sick_counts,
    "Deaths": death_counts
})

df.plot(kind = "bar", figsize = (8,6))
plt.title("COVID Cases and Deaths by Sex")
plt.ylabel("Count")
plt.xlabel("Sex")
plt.xticks(rotation = 0)
plt.show()

# Fatality
fatality_rate = (death_counts / sick_counts) * 100

fatality_rate.plot(kind = "bar", color = ["steelblue", "salmon"], figsize = (6,5))
plt.title("COVID Fatality Rate by Sex")
plt.ylabel("Fatality Rate (%)")
plt.xlabel("Sex")
plt.xticks(rotation = 0)
plt.show()

print(fatality_rate)
```



Based on the statistics, although more men (209,489) than women (182,490) were infected with COVID-19, the fatality rate was also higher among men (≈16.8%) compared to women (≈10.4%), indicating that men were more severely affected by COVID-19 deaths.

### Question 4: How many of the patients been in Intensive Care Unit, are still alive?

```python
been_in_icu = covid_data[covid_data.icu == "Yes"][["icu", "date_died"]]
alive_been_in_icu = been_in_icu[been_in_icu["date_died"].isna()]
result = (len(alive_been_in_icu) / len(been_in_icu)) * 100
print(f"{result:.2f}% of the patients who have been in ICU, are still alive.")
```

This statistic indicates that quite a lot of people with complications caused by COVID and / or other pre-conditions, could not be saved.
